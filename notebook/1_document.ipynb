**Document** - It is a type of a data structure provided by langchain.

It consists of two main field:
- metadata
- page_content

```python
from langchain_core.documents import Document
```

Importing a single text file

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text/AI.txt", encoding="utf-8")
document = loader.load()

print(document)
```

Importing from directory

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data/text",
    glob="**/*.txt" ,# patterns to match all the .txt files
    loader_cls = TextLoader,
)

documents = dir_loader.load()
print(documents)
```

Importing PDFs

```python
from langchain_community.document_loaders import PyMuPDFLoader

dir_pdf_loader = DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf" ,# patterns to match all the .txt files
    loader_cls = PyMuPDFLoader,
)

docu = dir_pdf_loader.load()
print(docu)
```
